# Tagging

```python
from pydantic import BaseModel, ValidationError, PositiveInt, confloat
from enum import Enum

"""
tagging_kwargs defaults:
* 'model': 'gpt-4o-mini'
* 'temperature': 0
* 'max_tokens': 30
* 'tag_threshold': 0.3 (tags that appear in >= 30% of the tagged chunks will be kept)
* 'untagged_option': 'keep_untagged'. There are 3 valid values:
    * 'keep_untagged': Keep untagged chunks and leave them untagged (empty list). Good for debugging.
    * 'discard_untagged': Discard untagged chunks
    * 'tag_untagged': Tag the untagged chunks with the same tags as all other chunks
"""
```

```python
class UntaggedOption(str, Enum):
    KEEP = 'keep'
    DISCARD = 'discard'
    TAG = 'tag'

class TaggingKwargs(BaseModel):
    model: str = 'gpt-4o-mini'
    temperature: float = 0
    max_completion_tokens: int = 30
    tag_threshold: confloat(ge=0, le=1) = 0.3
    untagged_option: UntaggedOption = UntaggedOption.KEEP
```

```python
# tagging_kwargs = TaggingKwargs(**{})
tagging_kwargs = TaggingKwargs(**{"untagged_option": "KEEP"})
print(tagging_kwargs)
# dir(tagging_kwargs.untagged_option)
# print(type(tagging_kwargs.untagged_option.name))
# print(tagging_kwargs.untagged_option.value)
# print(tagging_kwargs.model_dump())
# tagging_kwargs = TaggingKwargs(**tagging_kwargs.model_dump())
# tagging_kwargs
```

# Splitting

```python
from pydantic import BaseModel, ValidationError, PositiveInt, confloat
from enum import Enum
```

```python
split_strategy: SplitStrategy = SplitStrategy.fixed
split_strategy: SplitStrategy = SplitStrategy.recursive

class SplitStrategy(str, Enum):
    fixed = 'fixed'
    recursive = 'recursive'

class FixedSplittingKwargs(BaseModel):
    split_by: str = "word"
    split_length: int = 250
    split_overlap: int = 50
    split_threshold: int = 30
    respect_sentence_boundary: bool = True

class RecursiveSplittingKwargs(BaseModel):
    split_length: int = 250
    split_overlap: int = 50
    split_unit: str = "word"
    separators: list[str] = ["\f", "\n\n", "sentence", "\n", " "]
```
